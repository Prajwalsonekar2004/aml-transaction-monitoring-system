```python
df_account['Bank Name'] = df_account['Bank Name'].str.strip()
```

```python
df_account['Account Number'] = df_account['Account Number'].astype(str).str.strip().str.upper()
```

```python
df_account['Entity ID'] = df_account['Entity ID'].astype(str).str.strip().str.upper()
```

```python
df_account['Entity type'] = df_account['Entity Name'].str.extract(r'(^[A-Za-z ]+)')

df_account['Entity Number'] = df_account['Entity Name'].str.extract(r'(\d+)$')
```

```python
df_account['Account_type'] = df_account['Account Number'].apply(lambda x: 'Alphanumeric' if any(c.isalpha() for c in x) else 'Numeric')
```
